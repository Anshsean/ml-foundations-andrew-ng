This is a simulation of linear regression.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Notation used 
x train for input variable 
y train for output variable target 

```python
x_train = np.array([1.0,2.0])
y_train = np.array([300.0,500.0])
print(f"The input is {x_train}")
```

```python
x_train.shape
print(f"The number of training examples is {x_train.shape}")
```

```python
i = 1
x_i = x_train[i]
y_i = y_train[i]
print(f"(x^({i}),y^({i})) = ({x_i},{y_i})")
```

Scatter plot 

```python
plt.scatter(x_train,y_train,marker = 'x',c='r')
plt.title("Housing Prices")
plt.xlabel("Size")
plt.show()
```

```python
def compute_model_output(x,w,b):
    m = x.shape[0]
    f_wb = np.zeros(m) # where m is the number of training examples
    for i in range(m):
        f_wb[i] = w*x[i]+b
    return f_wb
```

```python
tmp_f= compute_model_output(x_train,200,100)

plt.plot(x_train,tmp_f,c='b',label = "Our Prediction")
plt.scatter(x_train,y_train,marker='x',c='r')
plt.legend
```

The required and most appropriate value of w and b are 200 and 100
