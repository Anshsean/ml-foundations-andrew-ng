```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
x_train = np.array([1.0,2.0])
y_train = np.array([1.5,2.5])
plt.scatter(x_train,y_train)
plt.plot(x_train,y_train)
```

```python
w = 1
b = 0.5
```

```python
x_test = x_train
y_test = np.zeros(2)
```

```python
def compute_output(w,b,x):
    m = x.shape[0]
    y_test = np.zeros(m)
    i=0
    for i in range(m):
        y_test = w*x+b
    return y_test

```

```python
y_test = compute_output(w,b,x_test)
plt.plot(x_test,y_test,c='r')
plt.scatter(x_train,y_train)
```

```python
def cost_function(y_test,y_train,m):
    sum = 0
    for i in range(m):
        error_sq = (y_test[i]-y_train[i])**2
        sum+=error_sq
    return sum/(2*m)
```

```python
cost_function(y_test,y_train,2)
```
